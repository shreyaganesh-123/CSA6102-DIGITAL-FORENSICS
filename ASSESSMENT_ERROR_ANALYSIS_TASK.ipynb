```python
custody_log = []

def add_entry(evidence_id, handler, action):
    entry = {
        'evidence_id': evidence_id,
        'handler': handler,
        'action': action
    }
    custody_log.append(entry)

def print_log():
    for entry in custody_log:
        print(entry['evidence_id'] + ' - ' + entry['handler'] + ' - ' + entry['action'])

add_entry('EVD-101', 'A.Kumar', 'Collected')
add_entry('EVD-101', 'S.Rao', 'Analysed')
print_log()
```

Q2

```python
def classify_incident(description):
    categories = {
        'Phishing': ['fake email', 'login link', 'credential'],
        'Ransomware': ['encrypted files', 'ransom note', 'bitcoin'],
        'Identity Theft': ['stolen identity', 'aadhaar', 'pan card']
    }

    description = description.lower()  # fix: call the method

    for category, keywords in categories.items():  # fix: use .items()
        for word in keywords:
            if word in description:
                return category

    return 'Unclassified'


result = classify_incident('Victim received a fake email with a login link')
print(result)
```

```python
import re

def is_suspicious(url):
    ip_pattern = '\\d+\\.\\d+\\.\\d+\\.\\d+'
    suspicious_words = ('verify', 'update', 'secure', 'login')

    if re.search(ip_pattern, url):
        return False
    for word in suspicious_words:
        if word not in url:
            return False
    return True

urls = ['http://192.168.1.10/verify-account', 'https://mybank.com/login']
for u in urls:
    print(u, '->', is_suspicious(u))
```

```python
def detect_bruteforce(log_lines, threshold=5):
    ip_counts = {}

    for line in log_lines:
        if 'FAILED LOGIN' in line:
            parts = line.split(' ')
            ip = parts[-1]  # last element is IP

            # initialize if not present
            ip_counts[ip] = ip_counts.get(ip, 0) + 1

    flagged = []
    for ip, count in ip_counts.items():  # fix: use .items()
        if count > threshold:
            flagged.append(ip)

    return flagged


log = [
    'FAILED LOGIN user=admin from 10.0.0.5',
    'FAILED LOGIN user=admin from 10.0.0.5',
    'FAILED LOGIN user=root from 10.0.0.5'
]

print(detect_bruteforce(log, threshold=2))
```

```python
import hashlib

def compute_hash(filepath):
    hasher = hashlib.sha256()

    with open(filepath, 'rb') as f:   # binary mode
        while True:
            chunk = f.read(4096)
            if not chunk:
                break
            hasher.update(chunk)

    return hasher.hexdigest()   # return hash string


def verify_integrity(filepath):
    hash_at_collection = compute_hash(filepath)
    hash_before_analysis = compute_hash(filepath)

    if hash_at_collection == hash_before_analysis:
        print('Match: Evidence integrity verified')
    else:
        print('Mismatch: Evidence may be compromised')


verify_integrity('evidence_image.dd')
```

```python
volatile_sources = ['RAM', 'Cache', 'Running Processes', 'Network Connections']

def classify(source):
    if source in volatile_sources:   # fixed: membership check
        return 'Volatile'
    else:
        return 'Non-Volatile'


evidence_sources = ['RAM', 'Hard Disk', 'Cache', 'Registry Files']

for src in evidence_sources:
    print(src, ':', classify(src))
```
