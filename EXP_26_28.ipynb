EXP 26

```python
import re

email = """
From: support@bank.com

Dear User,

Click here to verify your account:
https://fake-bank-login.com

Thank you.
"""

urls = re.findall(r'https?://\S+', email)

print("URLs Found:")
for url in urls:
    print(url)

    if "fake" in url:
        print("Status: Suspicious")
    else:
        print("Status: Legitimate")
```

EXP 27

```python
import hashlib

text = "This is a sample file."

data = text.encode()

print("MD5    :", hashlib.md5(data).hexdigest())
print("SHA1   :", hashlib.sha1(data).hexdigest())
print("SHA256 :", hashlib.sha256(data).hexdigest())
```

```python
files = ["photo.jpg", "report.pdf", "data.locked", "notes.encrypted"]

for file in files:
    if file.endswith(".locked") or file.endswith(".encrypted"):
        print(file, "-> Ransomware Indicator")
    else:
        print(file, "-> Safe")
```
